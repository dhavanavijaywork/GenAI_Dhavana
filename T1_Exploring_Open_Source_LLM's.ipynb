```python
from transformers import pipeline
```

```python
generator = pipeline('text-generation', model='gpt2')
```

```python
output = generator("Artificial intelligence is", max_new_tokens = 20)
```

```python
print(output)
```

```python
output = generator("Artificial intelligence is", min_new_tokens = 20)
```

```python
print(output)
```

```python
output = generator("The future of AI is", min_new_tokens = 20)
```

```python
print(output)
```

```python
output = generator("The capital of Karnataka is", max_new_tokens = 20)
```

```python
print(output)
```

```python
output = generator("A robot entered the classroom and", max_new_tokens = 20, temperature = 0.2, do_sample = True)
print(output)
```

```python
output = generator("A robot entered the classroom and", max_new_tokens = 20, temperature = 1.0, do_sample = True)
print(output)
```

```python
output = generator("A robot entered the classroom and", max_new_tokens = 20, temperature = 1.2, do_sample = True)
print(output)
```

```python
output = generator("A robot entered the classroom and", max_new_tokens = 20, temperature = 1.2, do_sample = False)
print(output)
```

```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```python
text = "My name is Arjun, I study at NIE."
```

```python
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
token_ids = tokenizer.encode(text)
print(token_ids)
```
